```python
# Python and Pandas Assessment Notebook

import pandas as pd
import numpy as np
from functools import reduce
```

```python
# Task 1: Create a DataFrame using Pandas
# Data dictionary setup
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']
}

# Create DataFrame and explicitly check data types
df = pd.DataFrame(data)
df['Age'] = df['Age'].astype(int)  # Enforce integer type

print("--- Task 1 Output ---")
print(df)
print("\nData Types:")
print(df.dtypes)
```

```python
# Task 2: Row and Column Manipulation
try:

    df_dropped = df.drop(columns=['City'])


    print("DataFrame after dropping 'City':")
    print(df_dropped)

    # Verification assertion
    assert 'City' not in df_dropped.columns, "City column was not dropped successfully!"
    print("Verification Passed: 'City' column is removed.")
except KeyError as e:
    print(f"Column error: {e}")
```

```python
# Task 3: Handling Null Values
# DataFrame with missing values
df_null = pd.DataFrame({
    'Name': ['Alice', None, 'Charlie', 'David'],
    'Age': [25.0, 30.0, np.nan, 28.0],
    'Score': [85, 90, 95, np.nan]
})


print(df_null)

# Method 1: Fill nulls per column type
df_filled = df_null.copy()
df_filled['Name'] = df_filled['Name'].fillna('Unknown')
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())  # Fill numeric with mean
df_filled['Score'] = df_filled['Score'].fillna(0)                     # Fill numeric with constant


print(df_filled)

# Method 2: Demonstrate dropping rows with nulls
df_dropped_nulls = df_null.dropna()
print("\nAlternative Method (dropna):")
print(df_dropped_nulls)
```

```python
# Task 4: GroupBy and Describe
df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]
})

grouped_desc = df_group.groupby('Category')['Value'].describe()
print(grouped_desc)

"""
Interpretation:
- Category 'A': Contains 3 items with a mean of 18.33, min of 10, max of 30, and std dev of 10.41.
- Category 'B': Contains 2 items with a mean of 22.50, min of 20, and max of 25.
- Category 'C': Contains 1 item with value 35.00 (std dev is NaN/0 due to single entry).
"""
```

```python
# Task 5: Concatenation and Merging
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10, 11, 12], 'D': [13, 14, 15, 16]})

# Concatenate df1 and df2 vertically
df_concat = pd.concat([df1, df2], axis=0, ignore_index=True)

# Merge/join horizontally with df3 on index
df_final = pd.merge(df_concat, df3, left_index=True, right_index=True)

print(df_final)
print("Shape of final DataFrame:", df_final.shape)
```

```python
# Task 8: Functions and Lambda
def square_number(num):
    if not isinstance(num, (int, float)):
        raise TypeError("Input must be a number.")
    return num ** 2

# Equivalent lambda function
square_lambda = lambda num: num ** 2 if isinstance(num, (int, float)) else "Invalid input"

inputs = [4, -3, 2.5]


for x in inputs:
    print(f"Input: {x} | Regular Function Result: {square_number(x)} | Lambda Result: {square_lambda(x)}")
```

```python
# Task 9: Iterators and Generators
# Custom Iterator Class
class FirstFiveEvens:
    def __init__(self):
        self.current = 0
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= 5:
            raise StopIteration
        val = self.current
        self.current += 2
        self.count += 1
        return val

# Generator Function
def even_generator():
    num = 0
    for _ in range(5):
        yield num
        num += 2


print("Iterator output:", list(FirstFiveEvens()))
print("Generator output:", list(even_generator()))
```

```python
# Task 10: Map, Reduce, and Filter
numbers = [1, 2, 3, 4, 5]

# map: square all numbers
squared_numbers = list(map(lambda x: x ** 2, numbers))

# reduce: product of all numbers
product_of_numbers = reduce(lambda x, y: x * y, numbers) if numbers else 0

# filter: keep only even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Original List:", numbers)
print("Squared Numbers (Map):", squared_numbers)
print("Product of Numbers (Reduce):", product_of_numbers)
print("Even Numbers (Filter):", even_numbers)
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
class Rectangle:
    def __init__(self, length: float, width: float):
        if length <= 0 or width <= 0:
            raise ValueError("Length and width must be positive values.")
        self.length = length
        self.width = width

    def area(self) -> float:
        return self.length * self.width

    def perimeter(self) -> float:
        return 2 * (self.length + self.width)

rect1 = Rectangle(10, 5)
rect2 = Rectangle(4.5, 3.5)

print(f"Rectangle 1 (10x5)  -> Area: {rect1.area()}, Perimeter: {rect1.perimeter()}")
print(f"Rectangle 2 (4.5x3.5) -> Area: {rect2.area()}, Perimeter: {rect2.perimeter()}")
```

```python

# Task 12: Pandas Data Analysis
df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})



# 1. Average salary by department
avg_salary_dept = df_employees.groupby('Department')['Salary'].mean()
print("1. Average Salary by Department:")
print(avg_salary_dept)

# 2. Extract names of employees with salary > 60000
high_earners = df_employees[df_employees['Salary'] > 60000]['Name'].tolist()
print("\n2. Employees with Salary > 60,000:")
print(high_earners)

# 3. Add 'Bonus' column (10% of salary)
df_employees['Bonus'] = df_employees['Salary'] * 0.10
print("\n3. Updated Employee DataFrame with Bonus:")
print(df_employees)
```
